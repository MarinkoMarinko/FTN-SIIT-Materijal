У *Python*-у можемо изводити различите нумеричке операције над бројевима. Основне операције су: сабирање (+), одузимање (-), множење (*), делјење (/), целобројно делјење (//), остатак при делјењу (%) и степеновање (**)

**Пример 5-Основи нумерички оператори**

````python
a = 10
b = 3

# Сабирање
zbir = a + b

# Одузимање
razlika = a - b

# Множење
proizvod = a * b

# Дељење (реално)
deljenje = a / b #float

# Целобројно дељење
celobrojno = a // b

# Остатак при делјењу
ostatak = a % b

# Степеновање
stepen = 2**5

# Испис резултата
print("Suma:", zbir)
print("Razlika:", razlika)
print("Proizvod:", proizvod)
print("Deljenje:", deljenje)
print("Celobrojno deljenje:", celobrojno)
print("Ostatak:", ostatak)
print("Stepen:", stepen)
````

Резултат реалног дељења увек је децимално број, без обзира на то да ли су операнди цели бројеви или не. То значи да чак и када делимо два целобројна податка, добијена вредност биће реалан број са децималним записом. Децимални тип података је `float`

#### *Float*
*Float* представља реалне бројеве са децималним зарезом

````python
x = 5.7
y = 2.3

rezultat = x + y
print(rezultat)
print(type(rezultat))
rezultat=int(rezultat)# imeTipa(koju promenljivu hocu da konvertujem)
print(rezultat)
print(type(rezultat))
````

У овом задатку променљиве `x`и `y` су типа `float`. Када их саберемо, резултат такође постаје `float`.

Затим се резултат кастује у тип `int` помоћу функције `int()`. Кастовање значи да се експлицитно мења тип податка. Функција `int()` не заокружује број, већ само одсеца децимални део.

## Енторке(*Touple*)
Енторке су сложени типови података које садрже један или више других података сложених један за другим. Енторке се дефинишу навођењем елемената унутар обичних заграда.


````python
x = ("a", 2, True)
print(x)
````

Енторка је једнозначно одређена елементима које је чине, као и редоследом којим су ти елементи наведени

````python
print((1, 2) == (1, 2))
print((1, 2) == (2, 1))
````

## *Скупови*
Скупови су скупови у математичком смислу: неуређене колекције елемената у којима се сваки елемент може наћи само једном.
Скупови се дефинишу на исти начин као енторке, једино што се елементи наводе у витичастим заградама, а поновљени елемнти се ингоришу

````python
x={1,2,2,3,4}
print(x)
y={0,3,True, False}
print(y)
````

````python
x={(2,3), 2, False}
print(x)
````

За разлику од енторки скупови су неуређене структуре, те редослед навођења елемената не мења скуп

````python
print({1,2}=={2,1})# ispisace se True jer oba skupa sadrze iste elemnte, redoles nije bitan
````

### Додавање и уклањање елемената скупа


````python
z = {"a", "b", "c"}
z.add("d")  # Dodaje element "d" u skup. Skup sada glasi {"a", "b", "c", d}
print(z)
z.discard("a")  # Uklanja element "a" iz skupa. Skup sada glasi {"b", "c", "d"}
print(z)
````

#### Математичке операције над скуопвима
Скупови у *Python*-у су типови података намењени да описују математичке скупове, те стога подржавају и уобичајене операције: унију, пресек, разлике, симетричне разлике.

- *Унија скупова*: ``` z1 | z2``` или ```z1.union(z2)```.
  - Унија садржи све елементе које *припадају једном или другом* скупу.
- *Пресек скупова*: ```z1 & z2``` или ```z1.intersection(z2)```
  - Пресек садржи све елементе који *припадају и једном и другом* скупу.
- *Разлика скупова*: ```z1 - z2``` или ```z1.difference(z2)```
  - Разлика скупова садржи све елементе који *припадају првом, али не и другом* скупу.


## Листе
Попут енторки и листе су сложени типови података који настају слагањем елемената једног за другим.
Хетерогене структуре, односно различити типови елемената се могу појављивати унутар исте листе.Насупрот енторкама, листе су променљиви типови података (енгл. _mutable_): елементи се могу додавати на почетак или крај, уметати у средину или брисати, без ограничења.
Листе се дефинишу на исти начин као енторке, једино се обичне заграде мењају угластима.

### Индексирање листе
преко угластих заграда, индекси почињу од нуле(као и код осталих колекција), могуће је индексирање негативним бројевима.

````python
x=["1", "a", "cao", 56, True]
print(x[4])
x[4]=False
print(x)
````

### Додавање елемената листи

Елементи се могу додати у листу на више различитих начина: додавањем елемента на крај (функција `append()`), спајањем са другом листом (функција `extend()`) или уметањем елемента на одређено место на листи (функција `insert()`).

````python
x=[1,2,3,4]
y=[5,6,7]
x.append(y)
print(x)
print(x[4])
````

### Уклањања елемената
Елементи се могу уклањати из на различите начине: по вредности (функција `remove()`), по позицији (функција `pop()`)

````python
x=[1,2,3,4]
x.remove(2)
print(x)
x=[1,2,3,4]
x.pop(2)
print(x)
````

## Речници
Речник је хетерогена структура коју чине елементи које идентификујемо по кључу. Другим речима, речник скуп паров кључ->вредност, где кључ на јединствен начин идентификује вредност. Вредности могу бити произвољног типа(могу бити бројеви, текст, али могу бити и раѕличите колекције: листе, енторке..), док код кључева постоје одређена ограничења, једно од њих је да кључеви морају бити јединствени.
Речници се дефинишу истом синтаксом као скупови, помоћу витичастих заграда, разлика је у томе што су елементи речника парови типа кључ-вредност облика `кључ : вредност`

# Управљање током извршавања


## Условни искази
У *Python*-у, ток извршавања програма може се променити помоћу условних исказа. Условни искази служе да се провери да ли је одређени услов за извршавање кода испуњен или није и у зависности од тога и неки део кода који следи иза исказа ће се извршити.

#### *if*

Исказ `if` служи за проверу тачности одређених израза. Уколико је израз (услов) тачан део кода који следи ће се извршити и обрнуто.
```
if uslov:
    # blok koda koji se izvršava ako je uslov True
```

*Python* користи увлачење за означавање блокова кода који припадају условима. Дакле, све линије кода након иф исказа које су увучене односе се на део кода који ће се потенцијално извршити у зависноснти од вредности услова



````python
x = 5
if x > 3: # probajte da menjate uslov, npr x>7
    print("x je veće od 3")  # Ovaj red je uvučen i deo je if bloka
    x = x + 1                    # Ovaj red je takođe uvučen
print("Izvan uslova")# Ovaj red nije uvučen i nije deo if bloka
print(x)# Ovaj red takođe nije deo if bloka
````

#### *if-else*

У *Python*-у, *if* и *else* структура омогућава доношење одлука у програму. Ако је услов у *if* исказу тачан (True), извршава се код унутар *if* блока, а код унутар *else* блока се прескаче. Ако је услов у *if* нетачан (False), тада се прескаче код унутар *if* блока, а извршава код унутар *else* блока. Услов у *else* није потребан јер се извршава само ако је претходни услов био False.

```
if uslov:
    # kod koji ce se izvrsiti ukoliko je uslov zadovoljen
else:
    # kod za sve ostale slučajeve
```

````python
x = 2
if x > 3:
    print("x je veće od 3")  # Ovaj red je uvucen i osnodi se na if iskaz
    x = x + 1                  # Ovaj red takođe
else:
    print("x nije veće od 3")  # Ovaj red je uvučen i odnosi se na else
    x = x * 2                  # Ovaj red takođe
print("Izvan uslova")  # Ovaj red nije deo if ili else bloka
print(x)  # Ovaj red ispisuje novu vrednost x
````

#### *if-elif-else*

Структура *if-elif-else* омогућава доношење одлука са више услова. Ако услов у *if* исказу није тачан, програм проверава услове у *elif* блоковима. Може бити више *elif* блокова, а у сваком се ставља специфичан услов који се проверава. Ако ниједан од услова у *if* или *elif* није тачан, извршиће се код у *else* блоку, који не садржи услов и користи се као подразумевани случај када сви претходни услови нису задовољени.


````python
x = 3
if x > 5:
    print("x je veće od 5")  # Ovaj red je uvučen i odnosi se na if
    x = x + 1
elif x == 3:
    print("x je tačno 3")  # Ovaj red je uvučen i odnosi se na elif
    x = x * 2
else:
    print("x je manje od 3")  # Ovaj red je uvučen i odnosi se na else
    x = x - 1
print("Izvan uslova")  # Ovaj red nije deo if, elif ili else bloka
print(x)  # Ovaj red ispisuje novu vrednost x
````

## Петље

Петље се користе ради вишеструког извршавања дела кода одређени број пута или до испуњења неког задатог услова

### *for* петља
*for* петља извршава дати блок кода одређени број пута, дакле користи се уколико је унапред познат број понављања дела кода. петља представља структурирани начин понављања (итерације) који омогућава пролазак кроз елементе колекција као што су листе, низови, стрингови, скупови или резултати функције `range()`
```
for element in kolekcija:
    blok_koda
```
где елемент представља привремену променљиву која у свакој итерацију узима вредност следећег елемента из колекције  


````python
lista=["roze", True, 2]
for element in lista:
      print(element)
````

`range(start, stop, step)`- je funkcija koja generiše niz celih brojeva:

start — početna vrednost (uključena),

stop — krajnja vrednost (isključena),

step — korak između brojeva.

````python
for i in range(0, 10, 2): #promenljivu i ne treba inicijalizovati posebno niti
# inkrementirati jer ona svakako uzima redom vrednosti iz neke liste, rečnika...
      print(i)
````

### *while* петља
 *while* петља омогућава извршавање блока кода све док је одређени услов тачан. За разлику од *for* петље која се користи када знамо тачан број понављања(или треба проћи кроз све елемнте неке колкције, па поново знамо тачно колико елемната има колекција и колико пута ће се итерација извршити), *while* се користи када број итерација зависи од неког услова који може да се промени током извршавања.

```
while uslov:
    blok_koda
```

````python
x = 0 #promenljiva se mora inicijalizovati
while x < 5: # x=0,1,2,3,4- telo petlje se izvršava
    print("x je:", x)
    x += 1 #u svakoj iteraciji se promenljiva inkrementira
````

# Графички приказ функција

За исцртавање функције једне променљиве *y = f(x)* потребна су два вектора: вектор вредности независне променљиве *x* и вектор одговарајућих вредности функције *y* за свако *x*.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([ 0, 0.5, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5])
y = np.array([1, 2, 3, 2, 2, 2, 1.5, 2, 2, 2.5])
p1 = plt.plot(x, y)
plt.show()
````

Паjтон jе исцртао дате вредности *y* у датим тачкама *x* и аутоматски те тачке споjио линиjама како би график изгледао као функциjа. Дакле, за наjjедноставниjе цртање функциjа довољно jе да припремимо два вектора броjева. У први вектор стављамо вредности независно променљиве *x* (вредности *x*-ose), а у други вектор стављамо вриjедности функциjе у тим тачкама.

````python
x = np.arange(0, 2*np.pi, 0.01) # start, stop, step
#x = np.linspace(0, 2*np.pi, 100). # start, stop, broj tačaka
y = np.sin(x)
p1 = plt.plot(x, y)
plt.show()
````

Графичко приказивање функциjа више променљивих ће бити приказано на примеру
фунцкиjе коjа зависи од две променљиве. Рецимо да желимо да нацртамо функциjу  *y=f(x1,x2) = sin (x1) + cos (x2)*  на интервалу *x1,x2 ∈[−5,5]*.
У овом случаjу имамо две независно променљиве, па морамо припремити два скупа
података за променљиве *x1* и *x2* и jедан скуп података за вредност функциjе *y*. Такође, ови скупови података више не могу бити вектори jер морамо да израчунамо вредност функциjе *y* за сваку комбинациjу вредности променљивих *x1* и *x2* из жељеног домена. Дакле, вредности променљивих *x1, x2* и *y* складиштимо у матрицама. У наставку jе приказан пример *Паjтон* кода коjи црта наведену функциjу, као и њен графички приказ коjи се добиjе као резултат.

````python
x1v = np.arange(-5, 5, 0.01)
x2v = np.arange(-5, 5, 0.01)
x1, x2 = np.meshgrid(x1v, x2v)
f = np.sin(x1) + np.cos(x2)
fig = plt.figure()
ax = fig.add_subplot(111,projection='3d')
p1 = ax.plot_surface(x1, x2, f)
plt.show()
````
