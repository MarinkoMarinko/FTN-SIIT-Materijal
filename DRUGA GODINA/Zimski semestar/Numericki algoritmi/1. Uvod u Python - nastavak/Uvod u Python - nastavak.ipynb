## NumPy i Matplotlib biblioteke

NumPy i Matplotlib su osnovne biblioteke za naučno računanje i vizuelizaciju u Pythonu.

---

### NumPy – rad sa nizovima (arrays)

**NumPy** omogućava efikasno rukovanje velikim numeričkim podacima kroz višedimenzionalne nizove.


````python
### Kreiranje i osnovne operacije
import numpy as np

a = np.array([1, 2, 3])
b = np.array([[1, 2], [3, 4]])

print("a =", a)
print("Dimenzije:", a.ndim)
print("Oblik:", b.shape)
````

````python
# Tipovi podataka i indeksiranje
print("Tip elemenata:", a.dtype)
print("Prvi element:", a[0])
print("Podniz:", b[:, 1])   # druga kolona
````

````python
# Matematicke operacije
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Zbir:", x + y)
print("Skalarni proizvod:", x * y)
print("Kvadrat:", x ** 2)
print("Srednja vrednost:", np.mean(x))
````

````python
# Rad sa matricama
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("A:")
print(A)
print("B:")
print(B)

# Matrični proizvod
C = np.dot(A, B)
print("A · B =\n", C)

# Alternativno korišćenjem @ operatora
print("A @ B =\n", A @ B)

# Transponovanje i determinanta
print("Transponovana A:\n", A.T)
print("Determinanta A:", np.linalg.det(A))
````

````python
# Višedimenzionalni nizovi
mat = np.arange(1, 10).reshape(3, 3)
print(mat)
print("Transponovana matrica:\n", mat.T)
````

````python
# Uslovno indeksiranje
v = np.array([3, -1, 7, 0, 5])
print("Maska:", v > 2)
print("Elementi veci od 2:", v[v > 2])
print("Elementi matrice veci od 4:", mat[mat > 4])
````

````python
# Sumiranje i agregacija
print("Zbir svih elemenata:", np.sum(mat))
print("Zbir po kolonama:", np.sum(mat, axis=0))
print("Zbir po redovima:", np.sum(mat, axis=1))
print("Maksimum niza v:", np.max(v), "na indeksu", np.argmax(v))
````

## Matplotlib

Matplotlib je biblioteka za plotovanje. U ovom delu će biti predstavljen deo ove biblioteke `matplotlib.pyplot`, koji nudi slične mogućnosti kao MATLAB.

````python
import matplotlib.pyplot as plt
````

Plot 2D podataka

````python
# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)

# Plot the points using matplotlib
plt.plot(x, y_sin)
````

````python
y_cos = np.cos(x)

# Plot the points using matplotlib
plt.plot(x, y_sin)
plt.plot(x, y_cos)
plt.xlabel('x axis label')
plt.ylabel('y axis label')
plt.title('Sine and Cosine')
plt.legend(['Sine', 'Cosine'])
````

Moguće je plotovati raličite stvari unutar iste slike

````python
# Compute the x and y coordinates for points on sine and cosine curves
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Set up a subplot grid that has height 2 and width 1,
# and set the first such subplot as active.
plt.subplot(2, 1, 1)

# Make the first plot
plt.plot(x, y_sin)
plt.title('Sine')

# Set the second subplot as active, and make the second plot.
plt.subplot(2, 1, 2)
plt.plot(x, y_cos)
plt.title('Cosine')

# Show the figure.
plt.show()
````

Učitavamo paket i podatke o pokemonima pozivom metode `read_csv()`.

Vizualizujemo podatke - prvih 5 redova - pozivom metode `head()`.

````python
print(df.columns)
````

Koliko ima ukupno redova u podacima:

````python
# 1. način
num_rows = len(df)
print(num_rows)

# 2. način
num_rows, num_columns = df.shape
print(num_rows)
````

### Čitanje podataka

Da pristupimo tačno određenoj koloni (npr. kolona `Name`) koristimo `df['Name']` ili `df.Name`. Oba pristupa daju isti rezultat.

````python
print(df['Name'])
print('\n--------------------\n')
print(df.Name)
````

Ipak, ako pristupimo koloni koja ima specijalne karaktere ili razmak (npr. kolona 'Type 1') onda prvi pristup funkcioniše, ali drugi ne. Primer:

````python
print(df['Type 1'])
print('\n--------------------\n')
# print(df.Type 1) # SyntaxError se dobije
````

Dalje možemo da vidimo koliko pokemona ima u svakoj kategoriji `Type 1`, tako što pozovemo metodu `value_counts()`.

````python
type1_pokemon_count = df['Type 1'].value_counts()
print(type1_pokemon_count)
````

Možemo da preuzmemo tačan broj pokemona koji pripada vodi, tako što čitamo vrednost `Water`:

````python
water_pokemon_count = type1_pokemon_count['Water']
print(water_pokemon_count)
````

#### Indeksiranje redova
Kada želimo da dobijom određeni red iz podataka (npr. 3. red), koristimo `iloc[3]` nad `DataFrame` objektom.

````python
thrid_row = df.iloc[3]
print(thrid_row)
````

Kada želimo da dobijmo redove od 3. do 5. reda, koristimo sintaksu kao za *slicing* redova:

````python
rows = df.iloc[3:5]
rows.head()
````

Da iteriramo kroz sve redove postoji više načina. Prikazujemo dva učestala:

````python
# 1. način - već poznato iloc[] indeksiranje
for index in range(len(df)):
    row = df.iloc[index]
    print(row)
    print('---------')

# 2. način - poziv metode iterrows()
for index, row in df.iterrows():
    print(row)
    print('---------')
````

Ako želimo da iteriramo kroz tačno određene redove, npr. da iteriramo od 3. do 5. reda:

````python
for index in range(3, 5):
    row = df.iloc[index]
    print(row)
    print('---------')
````

### Čitanje redova prema uslovu
Umesto po indeksu, možemo da izdvojimo redove po nekom uslovu. Želimo redove svih vodenih pokemona (videli smo gore da ukupno takvih pokemona ima 112).

````python
# 1. način
water_pokemon = df[df['Type 1'] == 'Water']

# 2. način
water_pokemon = df.loc[df['Type 1'] == 'Water']
water_pokemon.head() # da vidimo prvih 5
````

Možemo napraviti komplikovaniji uslov, da dobijemo vodene pokemone i koji imaju zdravlje `HP` iznad 70 poena:

````python
water_pokemon_hp_gt_70 = df[(df['Type 1'] == 'Water') & (df['HP'] > 70)]
water_pokemon_hp_gt_70.head() # da vidimo prvih 5
````

*Napomena: naprednije indeksiranje radićemo u narednim terminima vežbi.*

### Sortiranje vrednosti
Možemo sortirati podatke. 

Sortiramo prema `Type 1` rastuće, a prema `HP` (*health points*) opadajuce:

````python
df_sorted = df.sort_values(['Type 1', 'HP'], ascending=[1,0])
df_sorted.head(100) # prvih 100 redova
````

### Pisanje novih podataka
Možemo dodati nove kolone. Pravimo kolonu `Total` koja procenjuje koliko vredi svaki pokemon prema nekoj formuli.

````python
df['Total'] = df['HP'] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def'] + df['Speed']
df.head()
````

### Brisanje podataka
Mozemo brisati redove ili kolone (ukoliko smatramo da za to ima potrebe) pozivom metode `drop`.

````python
# brisanje prva dva reda
df = df.drop([0, 1])

# brisanje kolone Type 2
df = df.drop(columns=['Type 2'])

df.head() # da vidimo prvih 5
````

### Čuvanje podataka
Finalno, sačuvamo naše modifikovane podatke pozivom metode `to_csv()`.

````python
df.to_csv('modified_pokemon_data.csv', index=False, sep=',')
````
